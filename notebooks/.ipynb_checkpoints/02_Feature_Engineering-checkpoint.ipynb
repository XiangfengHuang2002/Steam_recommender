```python
# 基础库
import pandas as pd
import numpy as np

# 可视化库（用于检查特征分布）
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.preprocessing import StandardScaler

# 加载清洗后的数据
input_file = "../data/steam_games_cleaned.csv"
df = pd.read_csv(input_file)
df.head()
```

```python
# 确定数值特征
numeric_cols = ['price', 'positive', 'negative', 'recommendations', 
                'average_playtime_forever', 'metacritic_score']

# 填充缺失值
for col in numeric_cols:
    df[col] = df[col].fillna(0)

# 检查数值列统计信息
df[numeric_cols].describe()
```

```python
# 对极偏态字段做 log1p 变换
skewed_cols = ['positive', 'negative', 'recommendations', 'average_playtime_forever', 'price']

for col in skewed_cols:
    df[f'log1p_{col}'] = np.log1p(df[col])

# 查看变换后的统计信息
df[[f'log1p_{c}' for c in skewed_cols]].describe()
```

```python
# 标签/类别字段
categorical_cols = ['tags', 'categories', 'genres', 'languages']

# 将空值填充为空字符串
for col in categorical_cols:
    df[col] = df[col].fillna("")
```

```python
# 总评价数
df['total_reviews'] = df['positive'] + df['negative']

# 好评率（百分比）
df['positive_percent'] = df.apply(
    lambda row: (row['positive']/row['total_reviews']*100) if row['total_reviews'] > 0 else 0, axis=1
)

# 加权评分（考虑数量和好评率）
df['weighted_score'] = df['positive_percent'] * np.log1p(df['total_reviews'])

# 是否免费
df['is_free'] = (df['price'] == 0).astype(int)

# 价格区间（离散化）
price_bins = [0, 5, 10, 20, 50, 100, 200, np.inf]
df['price_bin'] = pd.cut(df['price'], bins=price_bins)
```

```python
# 拼接文本特征，用于后续 embedding
df['text_features'] = (
    df['tags'] + " " + 
    df['categories'] + " " + 
    df['genres'] + " " + 
    df['short_description'].fillna('')
)

df['text_features'].head()
```

```python
# 标准化数值特征，用于后续模型或 embedding 融合
num_features = ['log1p_price','log1p_positive','log1p_negative','log1p_recommendations',
                'average_playtime_forever','weighted_score']
scaler = StandardScaler()
num_scaled = scaler.fit_transform(df[num_features])

# 保存标准化后的数值特征
df_features = pd.DataFrame(num_scaled, columns=[f'scaled_{c}' for c in num_features])
df_features.head()
```

```python
import joblib

# 保存处理后的 DataFrame
output_file = "../data/steam_games_features.csv"
df.to_csv(output_file, index=False, encoding='utf-8')

# 保存 StandardScaler 对象
scaler_file = "../data/num_scaler.pkl"
joblib.dump(scaler, scaler_file)

print("Feature engineering 完成，数据保存至：", output_file)
print("Scaler 保存至：", scaler_file)
```
