```python
import pandas as pd
import numpy as np
from sklearn.metrics.pairwise import cosine_similarity
from sentence_transformers import SentenceTransformer

# 数据路径
features_file = "../data/steam_games_features.csv"
df = pd.read_csv(features_file)
df = df.sort_values('positive', ascending=False).head(5000).reset_index(drop=True)
df.head()
```

```python
import joblib

scaler_file = "../data/num_scaler.pkl"
scaler = joblib.load(scaler_file)

# 如果需要重新标准化数值特征
num_features = ['log1p_price','log1p_positive','log1p_negative','log1p_recommendations',
                'average_playtime_forever','weighted_score']

df[num_features] = scaler.transform(df[num_features])
df.head()
```

```python
model = SentenceTransformer('all-MiniLM-L6-v2')  # 小模型，速度快
text_embeddings = model.encode(df['text_features'].tolist(), show_progress_bar=True)

print("文本 embedding 维度：", text_embeddings.shape)
```

```python
num_scaled = df[num_features].values
# 拼接数值 + 文本向量
features = np.hstack([text_embeddings, num_scaled])

print("最终特征矩阵维度：", features.shape)
```

```python
def recommend_game(game_name, df, features, top_k=5):
    if game_name not in df['name'].values:
        print(f"{game_name} 不在数据集中")
        return []
    
    idx = df[df['name'] == game_name].index[0]
    query_vector = features[idx].reshape(1, -1)
    
    # 计算余弦相似度
    sims = cosine_similarity(query_vector, features)[0]
    
    # 排序，排除自己
    top_idx = sims.argsort()[::-1][1:top_k+1]
    return df.iloc[top_idx][['name','genres','tags','weighted_score']]

# 测试
recommend_game("Stardew Valley", df, features, top_k=5)
```

```python
from sklearn.manifold import TSNE
import matplotlib.pyplot as plt
from matplotlib import rcParams

# 降维到2D
tsne = TSNE(n_components=2, random_state=42)
embedding_2d = tsne.fit_transform(features[:1000])  # 前1000个游戏，防止太慢

plt.rcParams['font.sans-serif'] = ['SimHei']  # 或者 'Microsoft YaHei'
plt.rcParams['axes.unicode_minus'] = False

plt.figure(figsize=(8,6))
plt.scatter(embedding_2d[:,0], embedding_2d[:,1], s=5)
plt.title("TSNE 可视化游戏 embedding")
plt.show()
```

```python
sample_games = ["Stardew Valley", "Cyberpunk 2077", "Among Us"]
for game in sample_games:
    print(f"\nTop 5 similar games to {game}:")
    print(recommend_game(game, df, features, top_k=5))
```
