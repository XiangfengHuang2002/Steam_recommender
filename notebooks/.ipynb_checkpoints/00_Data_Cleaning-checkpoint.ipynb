```python
import os
import json
import pandas as pd

input_file = "../data/games.json"
output_file = "../data/steam_games_cleaned.csv"

# 保留字段
fields = [
    'appID', 'name', 'required_age', 'price', 'short_description',
    'tags', 'categories', 'genres', 'languages', 'reviews', 'website',
    'positive', 'negative', 'recommendations',
    'average_playtime_forever', 'median_playtime_forever', 'header_image',
    'screenshots', 'movies', 'metacritic_score'
]

dataset = {}
if os.path.exists(input_file):
    with open(input_file, 'r', encoding='utf-8') as f:
        dataset = json.load(f)

cleaned = []
for appID, game in dataset.items():
    row = {}
    row['appID'] = appID
    row['name'] = game.get('name', '').strip()
    row['short_description'] = game.get('short_description', '').strip()
    
    # 列表字段
    row['tags'] = " ".join(game.get('tags', []))
    row['categories'] = " ".join(game.get('categories', []))
    row['genres'] = " ".join(game.get('genres', []))
    row['screenshots'] = " ".join(game.get('screenshots', []))
    row['movies'] = " ".join(game.get('movies', []))
    
    # 填充默认值
    row['required_age'] = game.get('required_age', 0)
    row['price'] = game.get('price', 0.0)
    row['languages'] = game.get('supported_languages', '')
    row['reviews'] = str(game.get('reviews', ''))
    row['website'] = game.get('website', '')
    row['positive'] = game.get('positive', 0)
    row['negative'] = game.get('negative', 0)
    row['recommendations'] = game.get('recommendations', 0)
    row['average_playtime_forever'] = game.get('average_playtime_forever', 0)
    row['median_playtime_forever'] = game.get('median_playtime_forever', 0)
    row['header_image'] = game.get('header_image', '')
    row['metacritic_score'] = game.get('metacritic_score', 0)

    # 排除非游戏内容
    exclude_keywords = ["software", "video", "trailer", "demo", 
                        "utilities", "education", "audio production", 
                        "video production", "design", "animation"]
    
    def is_non_game(text):
        text = text.lower()
        return any(kw in text for kw in exclude_keywords)
    
    if (is_non_game(row['categories']) 
        or is_non_game(row['genres']) 
        or is_non_game(row['tags']) 
        or is_non_game(row['name'])):
        continue

    # 核心字段检查，如果缺失就丢掉
    if row['name'] and row['short_description'] and row['tags'] and row['categories'] and row['genres']:
        cleaned.append(row)

df_cleaned = pd.DataFrame(cleaned)
df_cleaned.to_csv(output_file, index=False, encoding='utf-8')
print(f"清洗完成，保留 {len(df_cleaned)} 条有效数据")
```
